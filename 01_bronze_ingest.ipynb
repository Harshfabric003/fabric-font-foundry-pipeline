### **Calling Google Fonts Data using API Key and structuring it in a manner**

```python
import requests
import pandas as pd

API_Key = "API Key"
url = f"https://www.googleapis.com/webfonts/v1/webfonts?key={API_Key}"

response = requests.get(url)
data = response.json()

# The API returns a list of font items under "items"
fonts = data["items"]

# Flatten into a simple table: family, category, variants, subsets
rows = []
for f in fonts:
    rows.append({
        "family": f.get("family"),
        "category": f.get("category"),
        "variants": ",".join(f.get("variants", [])),
        "subsets": ",".join(f.get("subsets", [])),
        "version": f.get("version"),
        "last_modified": f.get("lastModified")
    })

pdf = pd.DataFrame(rows)
```

### **Creation of Data Frame and storing it in the Lakehouse as a Table.**

```python
df = spark.createDataFrame(pdf)
df.write.format("delta").mode("overwrite").saveAsTable("bronze_google_fonts")
```
